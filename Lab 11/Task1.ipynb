```python
import numpy as nm
import matplotlib.pyplot as mtp
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('Mall_Customers.csv')
df.head()
```

```python

df.columns = df.columns.str.strip().str.replace(' ', '_')

x = df[['Gender', 'Age', 'Annual_Income_(k$)', 'Spending_Score_(1-100)']].copy()
x['Gender'] = x['Gender'].map({'Male': 0, 'Female': 1})
```

```python
wcss_list = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(x)
    wcss_list.append(kmeans.inertia_)

wcss_list
```

```python
mtp.figure(figsize=(8, 5))
mtp.plot(range(1, 11), wcss_list, marker='o')
mtp.title('Elbow Method (No Scaling)')
mtp.xlabel('Number of clusters (k)')
mtp.ylabel('WCSS')
mtp.tight_layout()
mtp.show()
```

```python
kmeans_no_scale = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_no_scale = kmeans_no_scale.fit_predict(x)
```

```python

mtp.figure(figsize=(8, 6))
colors = ['blue', 'green', 'red', 'black', 'purple']
labels = ['Cluster 1', 'Cluster 2', 'Cluster 3', 'Cluster 4', 'Cluster 5']
for i in range(5):
    mtp.scatter(
        x.values[y_no_scale == i, 2],
        x.values[y_no_scale == i, 3],
        s=100, c=colors[i], label=labels[i]
    )
mtp.scatter(
    kmeans_no_scale.cluster_centers_[:, 2],
    kmeans_no_scale.cluster_centers_[:, 3],
    s=300, c='yellow', label='Centroid', edgecolors='black'
)
mtp.title('Customer Clusters (No Scaling) — Income vs Spending')
mtp.xlabel('Annual Income (k$)')
mtp.ylabel('Spending Score (1-100)')
mtp.legend()
mtp.tight_layout()
mtp.show()
```

```python
x_scaled = x.copy()
cols_to_scale = ['Gender', 'Annual_Income_(k$)', 'Spending_Score_(1-100)']
scaler = StandardScaler()
x_scaled[cols_to_scale] = scaler.fit_transform(x_scaled[cols_to_scale])

x_scaled
```

```python
wcss_scaled = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(x_scaled)
    wcss_scaled.append(kmeans.inertia_)
```

```python
mtp.figure(figsize=(8, 5))
mtp.plot(range(1, 11), wcss_scaled, marker='o')
mtp.title('Elbow Method (Partial Scaling — Age Unscaled)')
mtp.xlabel('Number of clusters (k)')
mtp.ylabel('WCSS')
mtp.tight_layout()
mtp.show()
```

```python
kmeans_scaled = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_scaled = kmeans_scaled.fit_predict(x_scaled)
```

```python
mtp.figure(figsize=(8, 6))
for i in range(5):
    mtp.scatter(
        x.values[y_scaled == i, 2],
        x.values[y_scaled == i, 3],
        s=100, c=colors[i], label=labels[i]
    )
mtp.title('Customer Clusters (Partial Scaling) — Income vs Spending')
mtp.xlabel('Annual Income (k$)')
mtp.ylabel('Spending Score (1-100)')
mtp.legend()
mtp.tight_layout()
mtp.show()
```

```python
print("Cluster sizes (No Scaling):")
print(pd.Series(y_no_scale).value_counts().sort_index())

print("Cluster sizes (Partial Scaling):")
print(pd.Series(y_scaled).value_counts().sort_index())

print("Cluster Centers (No Scaling) — original feature order: Gender, Age, Income, Spending:")
print(nm.round(kmeans_no_scale.cluster_centers_, 2))

print("Cluster Centers (Partial Scaling — scaled space):")
print(nm.round(kmeans_scaled.cluster_centers_, 2))
```
