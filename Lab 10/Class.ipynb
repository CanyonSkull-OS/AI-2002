```python
from sklearn import datasets
import pandas as pd
import random
from sklearn.utils import shuffle
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = datasets.load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
data.isnull().sum()
```

```python
data.describe()
```

```python
data.corr()
```

```python
data['species'] = iris.target
```

```python
iris.target
```

```python
data = shuffle(data, random_state=42)
data
```

```python
data.dropna()
```

```python
data
```

```python
data.info()
```

```python
data.hist(figsize=(10, 10))
plt.suptitle('Histograms of Iris Dataset Features')
plt.show()
```

```python
import seaborn as sns
sns.countplot(x='species', data=data)
plt.title('Count of Each Iris Species')
plt.show()
```

```python
sns.scatterplot(x='sepal length (cm)', y='sepal width (cm)', hue='species', data=data)
plt.title('Sepal Length vs Sepal Width by Species')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Iris Dataset Features')
plt.show()
```

```python
sns.pairplot(data, hue='species')
plt.suptitle('Pairplot of Iris Dataset Features by Species', y=1.02)    
plt.show()
```

```python
x = data.drop(columns=['species'])
y = data['species']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.5)
```

```python
x_train
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
model = lr.fit(x_train, y_train)
```

```python
results = model.predict(x_test)
results
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy = accuracy_score(y_test, results)
confusion_matrix(y_test, results)
```

```python
print(classification_report(y_test, results))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
model = dt.fit(x_train, y_train)
```

```python
results = model.predict(x_test)
results
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy = accuracy_score(y_test, results)
print(f"Accuracy: {accuracy:.2f}")
confusion_matrix(y_test, results)
```
