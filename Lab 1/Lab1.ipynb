```python
#Task 7
def repeatedNumbersOnce(list1 : list):
    result = []
    for num in list1:
        if num not in result:
            result.append(num)

    return result

list1 = [1,2,2,3,5,5,7,2,34,7,8,3]
result1 = repeatedNumbersOnce(list1)
result1
```

```python
#Task 6
def longestSubstring(s):
    max_length = 1
    current_length = 1
    longest_char = s[0]

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            current_length += 1
        else:
            if current_length > max_length:
                max_length = current_length
                longest_char = s[i - 1]
            current_length = 1

    if current_length > max_length:
        max_length = current_length
        longest_char = s[-1]

    longest_substring = longest_char * max_length
    return longest_substring, max_length

# Example usage
input_string = "WEEWEEWOOWOOOOOOOAAAAAAAAAAAA"
substring, length = longestSubstring(input_string)
print(f"Longest substring: '{substring}' with length {length}")
```

```python
#Task 5
def parse_Dictionary():
    upper = 0
    lower = 0
    digit = 0
    string = input("Enter String: ")
    Nchar = len(string)
    for char in string:
        if char.isupper():
            upper += 1
    for char in string:
        if char.islower():
            lower += 1
    for char in string:
        if char.isdigit():
            digit += 1

    dict1 = {
        "NofChar": Nchar,
        "NofUpper": upper,
        "NofLower": lower,
        "NofDigit": digit,
    }

    return dict1


dict1 = parse_Dictionary()
dict1

```

```python
#Task 4
def SalaryCalculator(basic_Salary):
    HRA = 0.1 * basic_Salary
    DA = 0.05 * basic_Salary

    total_Salary = basic_Salary + HRA + DA

    return total_Salary

salary = int(SalaryCalculator(10000))
print(f"Updated Salary is: {salary}")
    
```
