```python
#Task 1
import math

class Node:
    def __init__(self, value=None):
        self.value = value
        self.children = []
        self.minmax_value = None


class MinimaxAgent:
    def __init__(self, depth):
        self.depth = depth

    def formulate_goal(self, node):
        return "Goal reached" if node.minmax_value is not None else "Searching"

    def act(self, node, environment):
        goal_status = self.formulate_goal(node)
        if goal_status == "Goal reached":
            return f"Minimax value: {node.minmax_value}"
        else:
            return environment.compute_minimax(node, self.depth)


class Environment:
    def __init__(self, tree):
        self.tree = tree
        self.states = []

    def get_percept(self, node):
        return node

    def compute_minimax(self, node, depth, maximizing_player=True):
        
        print(f"Visiting Node: {node.value}, Depth: {depth}, Maximizing: {maximizing_player}")
        self.states.append(node.value)

        if depth == 0 or not node.children:
            print(f"Leaf Node Reached: {node.value}")
            return node.value

        if maximizing_player:
            value = -math.inf
            for child in node.children:
                child_value = self.compute_minimax(child, depth - 1, False)
                value = max(value, child_value)

            node.minmax_value = value
            print(f"MAX chooses value {value} at Node {node.value}")
            return value

        else:
            value = math.inf
            for child in node.children:
                child_value = self.compute_minimax(child, depth - 1, True)
                value = min(value, child_value)

            node.minmax_value = value
            print(f"MIN chooses value {value} at Node {node.value}")
            return value


def run_agent(agent, environment, start_node):
    percept = environment.get_percept(start_node)
    agent.act(percept, environment)



root = Node("Start")

zoneA = Node("Zone A")
zoneB = Node("Zone B")

root.children = [zoneA, zoneB]

A1 = Node(3)
A2 = Node(5)

B1 = Node(2)
B2 = Node(9)

zoneA.children = [A1, A2]
zoneB.children = [B1, B2]

depth = 2

agent = MinimaxAgent(depth)
environment = Environment(root)

run_agent(agent, environment, root)

print("\nVisited States:", environment.states)

print("\nFinal Minimax Values:")
print("Start:", root.minmax_value)
print("Zone A:", zoneA.minmax_value)
print("Zone B:", zoneB.minmax_value)
```

```python
#Task 3
#Part b
import math

class Node:
    def __init__(self, value=None):
        self.value = value
        self.children = []
        self.minmax_value = None


class AlphaBetaAgent:
    def __init__(self, depth):
        self.depth = depth

    def act(self, node, environment):
        return environment.alpha_beta(node, self.depth, -math.inf, math.inf)


class Environment:
    def __init__(self, tree):
        self.tree = tree
        self.visited = []
        self.pruned = []

    def alpha_beta(self, node, depth, alpha, beta, maximizing=True):

        print(f"Visiting: {node.value}, α={alpha}, β={beta}, Max={maximizing}")
        self.visited.append(node.value)

        if depth == 0 or not node.children:
            return node.value

        if maximizing:
            value = -math.inf
            for child in node.children:
                value = max(value, self.alpha_beta(child, depth-1, alpha, beta, False))
                alpha = max(alpha, value)

                if beta <= alpha:
                    print(f"PRUNED at node: {child.value}")
                    self.pruned.append(child.value)
                    break

            node.minmax_value = value
            return value

        else:
            value = math.inf
            for child in node.children:
                value = min(value, self.alpha_beta(child, depth-1, alpha, beta, True))
                beta = min(beta, value)

                # Pruning
                if beta <= alpha:
                    print(f"PRUNED at node: {child.value}")
                    self.pruned.append(child.value)
                    break

            node.minmax_value = value
            return value


root = Node("Root")

attack = Node("Attack")
defend = Node("Defend")
gather = Node("Gather")

root.children = [attack, defend, gather]

A1, A2 = Node("A1"), Node("A2")
D1, D2 = Node("D1"), Node("D2")
G1, G2 = Node("G1"), Node("G2")

attack.children = [A1, A2]
defend.children = [D1, D2]
gather.children = [G1, G2]

A1.children = [Node(5), Node(6)]
A2.children = [Node(7), Node(4)]
D1.children = [Node(3), Node(8)]
D2.children = [Node(6), Node(2)]
G1.children = [Node(1), Node(9)]
G2.children = [Node(4), Node(7)]


agent = AlphaBetaAgent(depth=3)
env = Environment(root)

result = agent.act(root, env)

print("\nFinal Decision Value:", result)
print("Visited Nodes:", env.visited)
print("Pruned Nodes:", env.pruned)

print("\nMinimax Values:")
print("Root:", root.minmax_value)
print("Attack:", attack.minmax_value)
print("Defend:", defend.minmax_value)
print("Gather:", gather.minmax_value)
```
