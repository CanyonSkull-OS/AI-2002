```python
#Task 1
import random

class SecurityAgent:
    def __init__(self):
        self.patching = []
        
    
    def check(self, component, status):
        if status == 'Vulnerable':
            self.patching.append(component)
            return f'{component} Vulnerable'
        else:
            return f'{component} is Safe'
        
    def patch(self, environment):
        for comp in self.patching:
            environment.compdict[comp] = 'Safe'
            print(f'{comp} patched successfully')



class Environment:
    def __init__(self):
        self.components = ['C1', 'C2', 'C3',
                           'C4', 'C5', 'C6', 
                           'C7', 'C8', 'C9']
        self.status = ['Safe', 'Vulnerable']

        self.compdict = {comp: random.choice(self.status) for comp in self.components}

    def display(self):
        print("---------Initial System Check---------")
        for key, value in self.compdict.items():
            print(f"{key}:{value}")
        print("---------System Check Complete--------\n")
        

def run_agent(agent, environment):
    environment.display()

    for component, status in environment.compdict.items():
        result = agent.check(component, status)
        print(result)
    print('\n')


    agent.patch(environment)


env1 = Environment()
agent1 = SecurityAgent()

run_agent(agent1, env1)

```

```python
#Task 2
import random

class Environment:
    def __init__(self):
        self.servers = ['S1', 'S2', 'S3', 'S4', 'S5']
        self.loads = ['Balanced', 'Overloaded', 'Underloaded']

        self.serverloads = {server: random.choice(self.loads) for server in self.servers}
        
    def display(self, title):
        print(f'--------{title} Load Status--------')
        for key, value in self.serverloads.items():
            print(f'{key}: {value}')



class LoadBalancerAgent:
    def __init__(self):
        self.list = []
    
    def ScanServers(self, environment):
        for server, load in environment.serverloads.items():
            if(load != 'Balanced'):
                self.list.append(server)

    def BalanceServers(self, environment):
        print('\n')
        for server in self.list:
            environment.serverloads.update({server: 'Balanced'})
            print(f'{server} load balanced')
        print('\n')



def run_agent(agent, environment):
    environment.display('Initial')

    agent.ScanServers(environment)

    agent.BalanceServers(environment)

    environment.display('Final')

env1 = Environment()
agent1 = LoadBalancerAgent()

run_agent(agent1, env1)





    
    
```

```python
#Task 3
import random

class Environment:
    def __init__(self):
        self.tasks = ['T1', 'T2', 'T3', 'T4', 'T5']
        self.status = ['Completed', 'Failed']

        self.backups = {task: random.choice(self.status) for task in self.tasks}

    def display(self, title):
        print(f'--------{title}--------\n')
        for key, value in self.backups.items():
            print(f'{key}: {value}')


class BackupManagementAgent:
    def __init__(self):
        self.failed_tasks = []

    def scan_tasks(self, environment):
        print('--------Scanning Backup Tasks--------\n')
        for task, state in environment.backups.items():
            if state == 'Failed':
                self.failed_tasks.append(task)
                print(f'{task} has Failed')
            else:
                print(f'{task} is Completed')

    def retry_backups(self, environment):
        for task in self.failed_tasks:
            environment.backups[task] = 'Completed'
            print(f'{task} retried and completed')


def run_agent(agent, environment):
    environment.display("Initial Backup Status")

    agent.scan_tasks(environment)

    agent.retry_backups(environment)

    environment.display("Final Backup Status")


env1 = Environment()
agent = BackupManagementAgent()

run_agent(agent, env1)
```

```python
# Task 4
import random

class Environment:
    def __init__(self):
        self.components = ['A','B','C','D','E','F','G','H','I']

        self.states = ['Safe', 'Low Risk Vulnerable', 'High Risk Vulnerable']

        self.systemstatus = {comp: random.choice(self.states) for comp in self.components}

    def display(self, title):
        print(f'--------{title} System Status--------')
        for comp, state in self.systemstatus.items():
            print(f'{comp}: {state}')

class SecurityAgent:
    def __init__(self):
        self.vulnerable_components = []

    def ScanSystem(self, environment):
        for comp, state in environment.systemstatus.items():
            if state == 'Safe':
                print(f'Component {comp} is secure.')
            else:
                print(f'Warning: Component {comp} has {state}.')
                self.vulnerable_components.append(comp)

    def PatchVulnerabilities(self, environment):
        print('\n')
        for comp in self.vulnerable_components:
            state = environment.systemstatus[comp]
            if state == 'Low Risk Vulnerable':
                environment.systemstatus[comp] = 'Safe'
                print(f'Component {comp} patched successfully.')
            elif state == 'High Risk Vulnerable':
                print(f'Component {comp} requires premium service to patch.')
        print('\n')

def run_agent(agent, environment):
    environment.display('Initial')
    agent.ScanSystem(environment)
    agent.PatchVulnerabilities(environment)
    environment.display('Final')

env1 = Environment()
agent1 = SecurityAgent()

run_agent(agent1, env1)
```

```python
#Task 5
import random

class Environment:
    def __init__(self):
        self.rooms = ['R101','R102','R103','R104','R105']
        self.nurse_stations = ['N1','N2']
        self.medicine_storage = ['MediRoom1','MediRoom2']
        self.patients = {room: random.choice(['MedicineA','MedicineB','MedicineC']) for room in self.rooms}
        self.staff_available = {station: random.choice([True, False]) for station in self.nurse_stations}

    def display(self, title):
        print(f'--------{title} Hospital Status--------')
        for room, med in self.patients.items():
            print(f'{room}: requires {med}')
        for station, available in self.staff_available.items():
            print(f'Nurse Station {station}: {"Available" if available else "Not Available"}')

class DeliveryRobotAgent:
    def __init__(self):
        self.tasks = []

    def ScanEnvironment(self, environment):
        for room, med in environment.patients.items():
            print(f'Task assigned: Deliver {med} to {room}')
            self.tasks.append({'room': room, 'medicine': med})

    def ExecuteTasks(self, environment):
        print('\n')
        for task in self.tasks:
            storage = random.choice(environment.medicine_storage)
            print(f'Picking up {task["medicine"]} from {storage}')
            print(f'Moving to {task["room"]}')
            print(f'Scanning patient ID in {task["room"]}... Verified')
            if random.choice([True, False]):
                station = random.choice(environment.nurse_stations)
                if not environment.staff_available[station]:
                    print(f'Alert: Staff not available at {station}, alerting other staff')
            print(f'{task["medicine"]} delivered to {task["room"]}\n')

def run_agent(agent, environment):
    environment.display('Initial')
    agent.ScanEnvironment(environment)
    agent.ExecuteTasks(environment)
    environment.display('Final')

env1 = Environment()
agent1 = DeliveryRobotAgent()

run_agent(agent1, env1)
```

```python
#Task 6
class Environment:
    def __init__(self):
        self.rooms = {
            'a': 'Safe', 'b': 'Safe', 'c': 'Fire',
            'd': 'Safe', 'e': 'Fire', 'f': 'Safe',
            'g': 'Safe', 'h': 'Safe', 'j': 'Fire'
        }
        self.path = ['a','b','c','d','e','f','g','h','j']

    def display(self, title):
        print(f'--------{title} Room Status--------')
        grid = ['a','b','c','d','e','f','g','h','j']
        for i, room in enumerate(grid):
            status = self.rooms[room]
            symbol = '🔥' if status == 'Fire' else '✅'
            print(f'{room}:{symbol}', end='  ')
            if (i+1)%3 == 0:
                print()
        print('\n')

class FirefightingRobot:
    def __init__(self):
        self.current_room = None

    def MoveAndExtinguish(self, environment):
        for room in environment.path:
            self.current_room = room
            status = environment.rooms[room]
            print(f'Moving to room {room}')
            if status == 'Fire':
                print(f'Fire detected in {room}! Extinguishing...')
                environment.rooms[room] = 'Safe'
            else:
                print(f'Room {room} is safe.')
            environment.display(f'Status after visiting {room}')

def run_agent(agent, environment):
    environment.display('Initial')
    agent.MoveAndExtinguish(environment)
    environment.display('Final')

env1 = Environment()
agent1 = FirefightingRobot()

run_agent(agent1, env1)
```
