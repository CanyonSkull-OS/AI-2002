```python
#Pratice Task 3
class BankAccount():
    def __init__(self, ab) -> None:
        self.__Balance = ab

    def Deposit(self, amount):
        self.__Balance += amount
        print(f"{amount} Deposited")

    def Withdraw(self, amount):
        if(self.__Balance > amount):
            self.__Balance -= amount
            print(f"{amount} Withdrawn")
        else:
            print("Balance Insufficient...")

    def get_balance(self):
        print(f"Balance: {self.__Balance}")

A1 = BankAccount(20000)
A2 = BankAccount(10000)
A3 = BankAccount(30000)

A1.get_balance()
print("\n")
A2.Deposit(30)
A2.Withdraw(5000)
A2.get_balance()
print("\n")

A3.Withdraw(5432)
A3.get_balance()
print("\n")

```

```python
#Practice Task 2
class Staff():
    def __init__(self, n, id, dep) -> None:
        self.Name = n
        self.ID = id
        self.Dept = dep

class Teacher(Staff):
    def __init__(self, n, id, dep, c, sal) -> None:
        super().__init__(n, id, dep)
        self.Course = c
        self.Salary = sal

    def Teach(self):
        print("Teaching...\n")

class Admin(Staff):
    def __init__(self, n, id, dep, r, wh) -> None:
        super().__init__(n, id, dep)
        self.Role = r
        self.WorkingHours = wh

    def Perform(self):
        print("Performing Tasks...\n")

class ResearchAssistant(Staff):
    def __init__(self, n, id, dep, rt, stp) -> None:
        super().__init__(n, id, dep)
        self.Research_Topic = rt
        self.Stipend = stp

    def Research(self):
        print("Researching...\n")


T1 = Teacher("Omer Shahid", 855, "CS", "DSA", 500000)
A1 = Admin("Alisha Jawaid", 234, "CS", "PF", 300000)
R1 = ResearchAssistant("Talha Khan", 157, "CY", "CALC", 120000)


T1.Teach()
A1.Perform()
R1.Research()
```

```python
#Practice Task 1
class Light():
    def __init__(self):
        self.light = False

    def TurnOn(self):
        self.light = True
        print("Light Turned on...\n")

    def TurnOff(self):
        self.light = False
        print("Light Turned off...\n")

    def Display(self):
        if(self.light == False):
            print("Light is Off\n")
        else:
            print("Light is On\n")


LivingRoomLight = Light()
BedroomLight = Light()


LivingRoomLight.TurnOn()

BedroomLight.TurnOn()

LivingRoomLight.Display()

BedroomLight.Display()


```

```python
#Task 1
class SecurityAgent():
    def __init__(self, id : int, name : str, status : bool):
        self.ID = id
        self.Name = name
        self.Status = status

    def display(self):
        print(f"ID: {self.ID} \nName: {self.Name} \nStatus: {self.Status}")
        

class FirewallAgent(SecurityAgent):
    def __init__(self, id, name, status):
        super().__init__(id, name, status)

    def monitor_traffic(self):
        print("Monitoring Traffic....\n")
 
class MalwareDetectionAgent(SecurityAgent):
    def __init__(self, id, name, status):
        super().__init__(id, name, status)

    def scan_files(self):
        print("Scanning Files....\n")

class AutomationAgent(SecurityAgent):
    def __init__(self, id, name, status):
        super().__init__(id, name, status)

    def run_automation(self):
        print("Running Automation....\n")







FW = FirewallAgent(98745, "Grok 3.2V", True)
MD = MalwareDetectionAgent(32687, "Claude Sonnet 4.5 Pro", True)
AT = AutomationAgent(28376, "DeepSeek R4 Neo", True)

FW.display()
FW.monitor_traffic()

MD.display()
MD.scan_files()

AT.display()
AT.run_automation()
```

```python
#Task 2
class Threat():
    def __init__(self, id, name, severity):
        self.threat_id = id
        self.name = name
        self.severity = severity

    def display(self):
        print(f"ID: {self.threat_id} \nName: {self.name} \nSeverity: {self.severity}")


class PhishingThreat(Threat):
    def __init__(self, id, name, severity):
        super().__init__(id, name, severity)
    
    def analyze_emails(self):
        print("Analyzing Emails...\n")

class RansomwareThreat(Threat):
    def __init__(self, id, name, severity):
        super().__init__(id, name, severity)

    def scan_files(self):
        print("Scanning Files...\n")


class BotnetThreat(Threat):
    def __init__(self, id, name, severity):
        super().__init__(id, name, severity)

    def detect_traffic(self):
        print("Detecting Traffic...\n")



PT = PhishingThreat(23423, "GulluGullu", 5)
RT = RansomwareThreat(23475, "EengaBoonga", 8)
BT = BotnetThreat(87326, "Inki Pinki Ponki", 2)

PT.display()
PT.analyze_emails()

RT.display()
RT.scan_files()

BT.display()
BT.detect_traffic()

```

```python
#Task 3
class ResponseAgent():
    def execute_response(self):
        print("Executing Response...\n")

class AlertAgent(ResponseAgent):
    def execute_response(self):
        print("Sending Notification...\n")

class BlockAgent(ResponseAgent):
    def execute_response(self):
        print("Blocking Malicious Activities...\n")

class RecoverAgent(ResponseAgent):
    def execute_response(self):
        print("Restoring Affected Systems...\n")

BaseA = ResponseAgent()
AA = AlertAgent()
BA = BlockAgent()
RA = RecoverAgent()

BaseA.execute_response()
AA.execute_response()
BA.execute_response()
RA.execute_response()

```

```python
#Task 5
class Student():
    def __init__(self, grade, name):
        self.__grade = grade
        self.__name = name

    def set_grade(self, grade):
        self.__grade = grade

    def get_grade(self):
        print(f"Grade: {self.__grade}\n")

    def display_info(self):
        print(f"Name: {self.__name} \nGrade: {self.__grade}\n")


s1 = Student("A", "Aaraiz")
s2 = Student("B-", "Omer")

s1.display_info()
s2.display_info()

s1.set_grade("A+")
s1.set_grade("A-")

s1.get_grade()
s2.get_grade()

    
```
